```
import numpy as np


massa_boi = 2.0
densidade_agua = 1000.0
g = 9.81
volume_total = 10E-3


peso_boi = massa_boi * g


def funcao_reducao(h):

    r = ((3 * volume_total) / (4 * np.pi))**(1/3)
    V_s = np.pi * (h**2) * (3 * r - h) / 3
    peso_agua_deslocada = densidade_agua * g * V_s
    return peso_agua_deslocada - peso_boi


def secante(funcao, x0, x1, tol=1e-5, max_iter=100):
    for i in range(max_iter):
        fx0 = funcao(x0)
        fx1 = funcao(x1)

        if fx1 - fx0 == 0:
            print("Parada, divisão por zero!")
            return None


        x_new = x1 - fx1 * (x1 - x0) / (fx1 - fx0)


        if abs(x_new - x1) < tol:
            return x_new

        x0, x1 = x1, x_new

    print("Não convergiu!")
    return None

h0 = 0.0  # altura inicial
h1 = 1.0  # nova estimativa

altura_h_molhada = secante(funcao_reducao, h0, h1)

if altura_h_molhada is not None:
    print(f"A altura da porção molhada da boia é: {altura_h_molhada:.5f} m")
else:
    print("Erro ao calcular a altura da porção molhada.")
```
