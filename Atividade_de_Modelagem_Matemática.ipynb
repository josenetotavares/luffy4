# Underflow

É um erro de cálculo que ocorre quando o equipamento
computacional não consegue representar um número
real extremamente baixo dentro do conjunto número dos
reais.

```
import sys

# Obtém o menor valor positivo representável para um número de ponto flutuante
min_float = sys.float_info.min
print("Menor float representável:", min_float)

# Tentativa de causar um underflow
underflow_result = min_float / 2
print("Resultado do underflow:", underflow_result)
```

# Overflow

É um erro de cálculo que ocorre quando o equipamento
computacional não consegue representar um número
muito grande num cálculo número dentro de reta real.

```
import sys

# Obtém o maior valor representável para um número de ponto flutuante
max_float = sys.float_info.max
print("Maior float representável:", max_float)

# Tentativa de causar um overflow
overflow_result = max_float * 2
print("Resultado do overflow:", overflow_result)
```

# Erro de arredondamento

É um erro que se constitui da soma e multiplicação de
números de ponto flutuante cujas somas e produtos são
arredondados em um número mais legível e reduzido em
tamanho.

```
import math

# Definir o valor de x
x = 1

# Número de termos na série de Taylor
num_termos = 5

# Calcular a série de Taylor para e^x usando truncamento
e_aproximado = sum(x**n / math.factorial(n) for n in range(num_termos))

# Valor exato de e^x (usando math.e para x = 1)
e_exato = math.e

# Calcular o erro de truncamento
erro_truncamento = abs(e_exato - e_aproximado)

# Exibir os resultados
print("Valor aproximado de e^1 usando truncamento:", e_aproximado)
print("Valor exato de e^1:", e_exato)
print("Erro de truncamento:", erro_truncamento)
```

# Erro Absoluto e Erro Relativo


Erro Absoluto

Eabs = aex - aaprox

Em que aex é o valor exato da grandeza considerada e
aaprox é o valor aproximado da
mesma grandeza numérica.

Erro Relativo

Erel = aex − aaprox
aex


em que aex é o valor exato da grandeza considerada e
aaprox é o valor aproximado da mesma grandeza.
Podemos observar que o erro relativo nos fornece mais
informações sobre a qualidade do erro que estamos cometendo
num determinado cálculo, uma vez que no erro absoluto não é
levada em consideração a ordem de grandeza do valor
calculado, enquanto no erro relativo esta ordem é contemplada.


```
import math

# Valor exato de pi
valor_exato = math.pi

# Valor aproximado de pi
valor_aproximado = 3.14

# Calcular erro absoluto
erro_absoluto = abs(valor_exato - valor_aproximado)

# Calcular erro relativo
erro_relativo = erro_absoluto / abs(valor_exato)

# Exibir os resultados
print("Valor exato de π:", valor_exato)
print("Valor aproximado de π:", valor_aproximado)
print("Erro absoluto:", erro_absoluto)
print("Erro relativo:", erro_relativo)
print("Erro relativo (percentual):", erro_relativo * 100, "%")
```

# Erro de Truncamento

Ocorre quando o equipamento computacional não consegue
comportar a quantidade de dados produzidos a partir de uma
função com uma quantidade de termos que excede a
capacidade do armazenamento de gerar aquela numeração
específica.

```
import math

# Definir o valor de x
x = 1

# Número de termos na série de Taylor
num_termos = 5

# Calcular a série de Taylor para e^x usando truncamento
e_aproximado = sum(x**n / math.factorial(n) for n in range(num_termos))

# Valor exato de e^x (usando math.e para x = 1)
e_exato = math.e

# Calcular o erro de truncamento
erro_truncamento = abs(e_exato - e_aproximado)

# Exibir os resultados
print("Valor aproximado de e^1 usando truncamento:", e_aproximado)
print("Valor exato de e^1:", e_exato)
print("Erro de truncamento:", erro_truncamento)
```

# Propagação de erros

Quando desenvolvemos ou utilizamos um procedimento
numérico para buscar a solução de um determinado problema,
normalmente o processamento envolve um número muito
grande de operações elementares. Assim, na maioria das
vezes, o erro cometido em uma operação isolada pode não ser
muito significativo para a solução do problema que estamos
tratando, mas, sim, é necessário estar ciente como os erros se
propagam quando tratamos com muitas operações no
processamento.
O crescimento subsequente de um erro gerado é chamado de
erro propagado.
Neste caso, é fundamental termos o conhecimento de como
estes erros estão se propagando, isto é, caso estejam se
acumulando a uma taxa crescente, dizemos que o erro é
ilimitado, e a sequência de operações é considerada instável.
Se, por outro lado, os erros estão se acumulando a uma taxa
decrescente, dizemos que o erro é limitado e, portanto, a
sequência de operações é considerada estável.

```
import math

# Valor do raio e sua incerteza
raio = 10
delta_raio = 0.1

# Calcular a área exata do círculo
area_exata = math.pi * raio**2

# Calcular o erro absoluto na área usando a propagação de erros
delta_area = 2 * math.pi * raio * delta_raio

# Mostrar os resultados
print("Raio:", raio)
print("Incerteza no raio (Δr):", delta_raio)
print("Área exata do círculo:", area_exata)
print("Erro absoluto na área (ΔA):", delta_area)
```
