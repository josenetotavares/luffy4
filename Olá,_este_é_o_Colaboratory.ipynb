```
import numpy as np


R = 140
L = 0.26
C = 25e-6
vm = 24
im = 0.15
tolerance = 1e-8


def func(f):
    omega = 2 * np.pi * f
    return im - (vm / np.sqrt(R**2 + (omega*L - 1/(omega*C))**2))


def secant_method(f0, f1, tol, max_iter=1000):
    f_prev = f0
    f_curr = f1
    for _ in range(max_iter):

        f_next = f_curr - func(f_curr) * (f_curr - f_prev) / (func(f_curr) - func(f_prev))

        if abs(f_next - f_curr) < tol:
            return f_next
        f_prev, f_curr = f_curr, f_next
    return f_curr


f0 = 50
f1 = 60


frequencia_f = secant_method(f0, f1, tolerance)
print(f"A frequência f encontrada é: {frequencia_f} Hz")
```
