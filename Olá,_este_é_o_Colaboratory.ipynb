```

def f(x):
    return 1/2 + x**2 - 5

def bisseccao(a, b, tol=1e-5, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError("A função deve mudar de sinal entre a e b")

    iter_count = 0
    while (b - a) / 2 > tol and iter_count < max_iter:
        c = (a + b) / 2
        if f(c) == 0:
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c
        iter_count += 1

    return (a + b) / 2

# Exemplo de uso:
a, b = 1, 3  # Escolha um intervalo onde a função muda de sinal
raiz = bisseccao(a, b)
print(f"A raiz encontrada é: {raiz}")
```

```

def f(x):
    return 1/2 + x**2 - 5

def bisseccao(a, b, tol=1e-5, max_iter=100, criterio_convergencia='tolerancia'):
    if f(a) * f(b) >= 0:
        raise ValueError("A função deve mudar de sinal entre a e b")

    iter_count = 0
    while True:
        c = (a + b) / 2
        if f(c) == 0:
            return c, iter_count

        if criterio_convergencia == 'tolerancia':
            if (b - a) / 2 <= tol:
                return c, iter_count
        elif criterio_convergencia == 'iteracoes':
            if iter_count >= max_iter:
                return c, iter_count

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        iter_count += 1

# Exemplo de uso:
a, b = 1, 3  # Escolha um intervalo onde a função muda de sinal
raiz, iteracoes = bisseccao(a, b, criterio_convergencia='tolerancia')
print(f"A raiz encontrada é: {raiz}")
print(f"Número de iterações: {iteracoes}")
```

```

import numpy as np
import matplotlib.pyplot as plt

# Define a função
def f(x):
    return 1/2 + x**2 - 5

# Implementa o método da bissecção
def bisseccao(a, b, tol=1e-5, max_iter=100, criterio_convergencia='tolerancia'):
    if f(a) * f(b) >= 0:
        raise ValueError("A função deve mudar de sinal entre a e b")

    iter_count = 0
    while True:
        c = (a + b) / 2
        if f(c) == 0:
            return c, iter_count

        if criterio_convergencia == 'tolerancia':
            if (b - a) / 2 <= tol:
                return c, iter_count
        elif criterio_convergencia == 'iteracoes':
            if iter_count >= max_iter:
                return c, iter_count

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        iter_count += 1

# Encontrar a raiz usando o método da bissecção
a, b = 0, 5  # Intervalo de 0 a 5
tol = 0.001
raiz, _ = bisseccao(a, b, tol=tol, criterio_convergencia='tolerancia')

# Gera valores para x
x = np.linspace(a, b, 400)
y = f(x)

# Cria o gráfico
plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r'$f(x) = \frac{1}{2} + x^2 - 5$', color='blue')
plt.axhline(0, color='black', linewidth=0.5)  # Linha horizontal no eixo x
plt.axvline(0, color='black', linewidth=0.5)  # Linha vertical no eixo y
plt.scatter(raiz, f(raiz), color='red', zorder=5, label=f'Raiz: {raiz:.4f}')  # Marca a raiz em vermelho
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.title('Gráfico da função $f(x) = \frac{1}{2} + x^2 - 5$ com a raiz encontrada')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```
