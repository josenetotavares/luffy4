# Atividade de Modelagem Matemática Com A Falsa Posição e NewTon-Raphson

```
import math

g = 9.8
c = 15.0
v_desejado = 35.0
t = 9.0

def func(m):
    return (g * m / c) * (1 - math.exp(-c * t / m)) - v_desejado

def posicao_falsa(func, a, b, tol=1e-6, max_iter=100):
    if func(a) * func(b) >= 0:
        print("Não há raiz ou há múltiplas raízes no intervalo")
        return None

    for i in range(max_iter):
        m = (a * func(b) - b * func(a)) / (func(b) - func(a))
        if abs(func(m)) < tol:
            return m, i + 1
        if func(a) * func(m) < 0:
            b = m
        else:
            a = m
    return m, max_iter

def newton_raphson(func, deriv_func, x0, tol=1e-6, max_iter=100):
    x = x0
    for i in range(max_iter):
        fx = func(x)
        fpx = deriv_func(x)
        if abs(fx) < tol:
            return x, i + 1
        x = x - fx / fpx
    return x, max_iter

def derivada_func(m):
    term1 = (g / c) * (1 - math.exp(-c * t / m))
    term2 = (g * m / c) * (math.exp(-c * t / m) * (c * t / m**2))
    return term1 + term2

m_pos_falsa, iter_pos_falsa = posicao_falsa(func, 50, 100)
m_newton, iter_newton = newton_raphson(func, derivada_func, 70)

print(f"Método de posição falsa: Massa = {m_pos_falsa:.6f} kg, Iterações = {iter_pos_falsa}")
print(f"Método de Newton-Raphson: Massa = {m_newton:.6f} kg, Iterações = {iter_newton}")
```
